```python
# Step 1: Import Pandas
import pandas as pd

# Step 2: Load the CSV file
# Replace 'orders.csv' with your file path
df = pd.read_csv('orders.csv')

# Preview the data
print("Original Data:")
print(df.head())

# Step 3: Total revenue per city
revenue_per_city = df.groupby('city')['amount'].sum().reset_index()
revenue_per_city = revenue_per_city.rename(columns={'amount': 'total_revenue'})
print("\nTotal Revenue per City:")
print(revenue_per_city)

# Step 4: Top 3 customers by spending
top_customers = df.groupby('customer_id')['amount'].sum().reset_index()
top_customers = top_customers.sort_values(by='amount', ascending=False).head(3)
top_customers = top_customers.rename(columns={'amount': 'total_spent'})
print("\nTop 3 Customers by Spending:")
print(top_customers)

# Step 5: Add week_number column derived from order_date
df['order_date'] = pd.to_datetime(df['order_date'])  # Ensure date is in datetime format
df['week_number'] = df['order_date'].dt.isocalendar().week
print("\nData with Week Number:")
print(df.head())

# Step 6: Filter orders where amount > 3000 and city = "Delhi"
filtered_orders = df[(df['amount'] > 3000) & (df['city'] == 'Delhi')]
print("\nFiltered Orders (amount>3000 & city='Delhi'):")
print(filtered_orders)

# Step 7: Export filtered results to another CSV
filtered_orders.to_csv('filtered_orders.csv', index=False)
print("\nFiltered data exported to 'filtered_orders.csv'")
```
