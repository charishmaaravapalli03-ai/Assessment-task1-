```python
import numpy as np

# Original data
data = np.array([4.1, 4.3, np.nan, 4.5, 4.4, np.nan, 4.6])

# 1. Replace NaN values with the mean of non-NaN values
mean_value = np.nanmean(data)         # mean ignoring NaN values
cleaned_data = np.where(np.isnan(data), mean_value, data)

# 2. Calculate mean and standard deviation (after cleaning)
mean_final = np.mean(cleaned_data)
std_final = np.std(cleaned_data)

# 3. Rolling average (window size = 3)
rolling_avg = np.convolve(cleaned_data, np.ones(3)/3, mode='valid')

# 4. Detect anomalies: value > mean + 2*std
threshold = mean_final + 2 * std_final
anomalies = cleaned_data > threshold

# 5. Final cleaned dataset ready for ML
final_dataset = cleaned_data
```

```python
step-by-Step Explanation (Plain English)
 Step 1: Replace NaN values

The array has missing values (np.nan).

First, we calculate the mean of only the valid numbers using np.nanmean().

Then we replace every NaN with that mean using np.where().

 Step 2: Calculate Mean & Standard Deviation

After filling NaNs, the data is now complete.

We compute:

Mean using np.mean()

Standard deviation using np.std()

Step 3: Rolling Average (Window = 3)

A rolling average smooths the signal.

We use np.convolve() with a window of 3 to compute the average of every consecutive 3 values.

Example:
(4.1 + 4.3 + 4.46)/3
(4.3 + 4.46 + 4.5)/3
… and so on.
 Step 4: Detect anomalies

If a value is greater than mean + 2×standard deviation, it is considered an anomaly.

We create a boolean array that tells which values are anomalies.

 Step 5: Produce cleaned dataset

The cleaned dataset is the NaN-filled array.

This dataset has:

No missing values

No anomalies (they are identified)

Ready for machine learning
```
