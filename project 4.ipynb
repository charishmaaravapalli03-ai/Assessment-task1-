```python
import pandas as pd

# Try reading with default comma separator
try:
    orders = pd.read_csv("orders.csv")
except:
    # If fails, try whitespace separator
    orders = pd.read_csv("orders.csv", sep="\s+")

# Strip any spaces from column names
orders.columns = orders.columns.str.strip()

# Preview the first few rows and columns
print("Columns Detected:", orders.columns.tolist())
print(orders.head())
```

```python
# Convert amount to numeric
orders['Unit_Price'] = pd.to_numeric(orders['Unit_Price'], errors='coerce')

# Convert order_date to datetime
orders['Order_Date'] = pd.to_datetime(orders['Order_Date'], errors='coerce')
```
